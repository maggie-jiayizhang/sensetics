```python
import numpy as np 
from copy import deepcopy
```

```python
L = 4000
L_half = L // 2
d = 100
seed = 0
```

```python
dist = np.arange(0, L_half+d-1, d)

np.random.seed(seed)
np.random.shuffle(dist)
disp_plus = deepcopy(dist)
print("+x locations:", disp_plus)
for i in range(len(disp_plus)):
    print(f"{disp_plus[i]}")
np.random.shuffle(dist)
disp_minus = deepcopy(dist)*-1
print("-x locations:", disp_minus)
for i in range(len(disp_minus)):
    print(f"{disp_minus[i]}")
```

```python
def get_displacement(disp, x0=0):
    new_disp = deepcopy(disp)
    new_disp = np.insert(new_disp, 0, x0)
    return np.diff(new_disp)
```

```python
dx = get_displacement(disp_minus)
for loc in dx:
    print(f"{loc}")
```
