```python
import numpy as np 
from copy import deepcopy
```

```python
L = 3000
L_half = L//2
d = 200
seed = 0
```

```python
dist_half = np.arange(0, L_half+d-1, d)
dist_full = np.unique(np.concatenate((dist_half, dist_half*-1)))

np.random.seed(seed)
np.random.shuffle(dist_full)

for i in range(len(dist_full)):
    print(f"{dist_full[i]}")
```

```python
def get_displacement(disp, x0=0):
    new_disp = deepcopy(disp)
    new_disp = np.insert(new_disp, 0, x0)
    return np.diff(new_disp)
```

```python
dx = get_displacement(dist_full, x0=dist_full[0])
for loc in dx:
    print(f"{loc}")
```
